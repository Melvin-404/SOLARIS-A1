# 01 - SoLEXS Level-1 FITS Data & Metadata Inspection (Stage 3)

**Project:** SOLARIS-A1  
**Objective:** Inspect raw Level-1 Aditya-L1 SoLEXS light-curve and GTI FITS files to extract actual HDU structures, column definitions, time-system conventions, and numerical data ranges without hardcoded assumptions.

```python
import sys
from pathlib import Path
import pandas as pd
from astropy.io import fits

# Ensure src/ is on sys.path
cwd = Path(".").resolve()
project_root = cwd.parent if cwd.name == "notebooks" else cwd
src_dir = project_root / "src"
if str(src_dir) not in sys.path:
    sys.path.insert(0, str(src_dir))

from solar_flare.inspection.fits_inspect import (
    list_hdus,
    read_header_dict,
    describe_columns,
    summarise_numeric_columns,
    summarise_time_columns,
)
from solar_flare.inspection.time_systems import extract_time_system_metadata

print(f"Project root resolved to: {project_root}")
```

```python
# Configuration: Select raw FITS file to inspect
target_fits = project_root / "data" / "raw" / "AL1_SOLEXS_20261018_SDD2_L1.lc.gz"
print(f"Target FITS file : {target_fits.name}")
print(f"File exists      : {target_fits.is_file()}")
print(f"File size        : {target_fits.stat().st_size:,} bytes")
```

## 1. HDU Structure Listing

```python
hdus = list_hdus(target_fits)
pd.DataFrame(hdus)
```

## 2. Table Column Descriptors (Extension HDU 1)

```python
columns_meta = describe_columns(target_fits, hdu=1)
pd.DataFrame(columns_meta)
```

## 3. Time System Keywords and Candidate Interpretations

```python
time_meta = extract_time_system_metadata(target_fits)
print("Observed Consolidated Time Keywords:")
for kw, val in time_meta["consolidated_keywords"].items():
    print(f"  {kw:10s} = {val}")

print("\nCandidate Interpretations (Requires Student Confirmation):")
for interp in time_meta["candidate_interpretations"]:
    print(f"  - {interp}")
```

## 4. Column Statistics and Diagnostics

```python
num_stats = summarise_numeric_columns(target_fits, hdu=1)
df_stats = pd.DataFrame(num_stats).T
display(df_stats)

time_diag = summarise_time_columns(target_fits, hdu=1)
display(pd.DataFrame(time_diag))
```

## 5. Table Data Sample (First 10 Rows)

```python
with fits.open(str(target_fits), mode="readonly", memmap=False) as hdul:
    table_data = hdul[1].data
    df_sample = pd.DataFrame(table_data[:10])

df_sample
```
